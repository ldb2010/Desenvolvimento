# Aula 1 - Importanto Pacotes e Tabela de Frequência

Vamos começar importando os pacotes necessários. 

```python
import pandas as pd
```

Importando a base de dados. Lembre que você precisa ter salvo o arquivo na sua pasta de trabalho. Lembre que você precisa indicar para o Python onde fica a sua pasta de trabalho. Faça isso do seguinte modo: File -> Open Folder. 

```python
base = pd.read_excel("Resposta_Musical.xlsx")
```

Verificar as 10 primeiras linhas
base.head(10)

```python
base.head(10)
```

Calcular a frequência absoluta

```python
genero = base["Genero_Musical"].value_counts()
```

Podemos verificar a estrutura dos dados

```python
genero
```

Colocando o total de participantes

```python
total = genero.sum()
print(total)
```

Quero transformar em um dataframe. Desse modo, podemos trabalhar com mais liberdade com os dados

```python
df = pd.DataFrame(genero)
df
```

Criar uma nova coluna com a frequência relativa. Lembre que a frequência relativa é dada pela seguinte relação $Frequência_Relativa = \frac{Frequência_Relativa}{Total}$ 

```python
df["Freq_Relativa"] = df["count"]/ total 
df
```

Criar uma coluna com para frequência relativa percentual.

```python
df["Freq_Relativa_%"] = df["Freq_Relativa"]*100
df
```

Trocar o nome das colunas.

```python
df.columns = ["Freq_Absoluta","Freq_Relativa","Freq_Relativa_%"]
df
```

```python
df.loc['Total'] = total
df.loc['Total', 'Freq_Relativa'] = 1.0
df
```

Posso querer saber o valor de algum dado específico.

```python
df.loc["Rap"]
```

Podemos exportar os dados tratados para o excel

```python
df.to_excel("teste2.xlsx")
```
