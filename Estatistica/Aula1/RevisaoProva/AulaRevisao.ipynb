```python
import pandas as pd
```

```python
base = pd.read_excel("Resposta_Musical.xlsx")
base
```

```python
genero = base["Genero_Musical"].value_counts()
genero
```

```python
total = genero.sum()
print(total)
```

```python
genero_df = pd.DataFrame(genero)
genero_df
```

Criar Frequencia Relativa

```python
genero_df["Frequencia Relativa"] = genero_df["count"]/total
genero_df["Frequencia Relativa %"] = genero_df["Frequencia Relativa"]/total
genero_df
```
