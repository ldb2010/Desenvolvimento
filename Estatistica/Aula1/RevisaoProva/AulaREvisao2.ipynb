```python
import pandas as pd
base = pd.read_excel("Atividade_Tabela_Freq.xlsx")
base
```

```python
rede = base["REDES"].value_counts()
rede
```

```python
total = rede.sum()
print(total)
```

```python
redes_df = pd.DataFrame(rede)
redes_df
```

```python
redes_df["Frequencia Relativa"] = redes_df["count"]/total
redes_df["Frequencia Relativa %"] = redes_df["Frequencia Relativa"]/total
redes_df
```
