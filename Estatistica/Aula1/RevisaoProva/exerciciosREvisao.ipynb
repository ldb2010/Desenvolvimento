```python
base = pd.read_excel("Atividade_Tabela_Freq.xlsx")
base.head(5)
```

Exercicio2-FA

```python
redes = base["REDES"].value_counts()
redes
```

Exercicio3-TotalREspostas

```python
total = redes.sum()
total
```

Exercicio4-FR%

```python
fr = (redes/total)*100
fr
```

Exercicio5-Tabela Completa

```python
df = pd.DataFrame(redes.reset_index().values, columns=["Rede Social","FA"])
df["FR"] = df["FA"]/total
df["FR%"] = df["FR"]*100
df
```
