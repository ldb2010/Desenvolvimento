```python
import pandas as pd
```

```python
rs = pd.read_excel("exercicio_prova_tabelas.xlsx")
redes = rs["REDES"].value_counts()
totalRedes = redes.sum()
redesDF = pd.DataFrame(redes.reset_index().values, columns=["Rede Social", "Frequencia Absoluta"])
redesDF["Frequencia Relativa"] = redesDF["Frequencia Absoluta"]/totalRedes
redesDF["FR%"] = redesDF["Frequencia Relativa"]*100
redesDF.loc["Total: "] = redesDF.sum()
redesDF.loc["Total: ", "Rede Social"] = "Total"
redesDF-
```

```python
g = pd.read_excel("exercicio_prova_tabelas.xlsx")
genero = g["Genero"].value_counts()
totalG = genero.sum()
generoDF= pd.DataFrame(genero.reset_index().values, columns=["Genero","FA"])
generoDF["FR"] = generoDF["FA"]/totalG
generoDF["FR%"] = generoDF["FR"]*100
generoDF.loc["Total: "]= generoDF.sum()
generoDF.loc["Total: ", "Genero"]="Total"
generoDF
```

```python
notas= pd.read_excel("exercicio_prova_tabelas.xlsx")
letra = notas["Letras"].value_counts()
totalN = letra.sum()
notasDF = pd.DataFrame(letra.reset_index().values,columns=["Notas","FA"])
notasDF["FR"] = notasDF["FA"]/totalN
notasDF["FR%"] = notasDF["FR"]*100
notasDF.loc["Total: "] = notasDF.sum()
notasDF.loc["Total: ","Notas"]= "Total"
notasDF
```
