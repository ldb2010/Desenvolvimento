# Aula 2 Atividade com a Tabela de Frequênca - Revisão 

Vou importar os pacotes necessários

```python
import pandas as pd
```

Vou importar o arquivo de excel que será utilizado 

```python
base = pd.read_excel("Atividade_Tabela_Freq.xlsx")
base
```

Vou calcular a frequência absoluta.

```python
redes = base["REDES"].value_counts()
redes
```

Calcular o total

```python
total = redes.sum()
total
```

Transformar em uma base de dados.

```python
df = pd.DataFrame(redes.reset_index().values, columns=['Rede Social', 'Frequência Absoluta'])
df
```

Calcular a frequência relativa

```python
df["Frequência Relativa"] = df["Frequência Absoluta"]/total 
df
```

```python
df["Frequência Relativa Percentual"] = df["Frequência Relativa"]*100 
df
```

```python
df.loc["Total"] = df.sum()
df
```

```python
df.loc["Total","Rede Social"] = "Total"
df
```

Agora vou fazer o mesmo procedimento para turmas. Vou fazer de um modo mais fácil nesse caso. 



```python
base
turma = base["Turma"].value_counts()
turma
```

```python
turma_df["FR%"] = turma_df["FR"]*100
turma_df
```
