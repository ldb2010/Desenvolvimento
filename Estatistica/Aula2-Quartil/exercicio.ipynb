```python
import plotly_express as px
```

```python
import pandas as pd
```

```python
dados = [-30,1,2,3,4,12,14,15,16,42]
```

```python
df = pd.DataFrame(dados)
df.columns = ["Numero"]
df
```

```python
q1 = df["Numero"].quantile(0.25)
q2 = df["Numero"].quantile(0.50)
q3 = df["Numero"].quantile(0.75)

diq = q3 - q1
```

```python
if df["Numero"].max()>q3+1.5*diq:
    ls = q3+1.5*diq
    print("Caso 1")
else:
    ls = df["Numero"].max()
    print("Caso 2")
```

```python
if df["Numero"].min()<q1+1.5*diq:
    li = q3-1.5*diq
    print("Caso 1")
else:
    li = df["Numero"].min()
    print("Caso 2")
```

```python
figura1= px.box(y=df["Numero"])
figura1.show()
```

```python
total = q1+q2+q3+diq+ls+li
print(total)
```
