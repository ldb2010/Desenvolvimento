```python
import pandas as pd
```

```python
import plotly_express as px
```

```python
base_A = pd.read_excel("Dados_Box.xlsx",sheet_name="TurmaA")
```

```python
q1A = base_A["Nota"].quantile(0.25)
q2A = base_A["Nota"].quantile(0.50)
q3A = base_A["Nota"].quantile(0.75)

diq = q3A - q1A
```

```python
if base_A["Nota"].max()<q3A+1.5*diq:
    ls = q3A+1.5*diq
else:
    ls = base_A["Nota"].max()
```

```python
if base_A["Nota"].min()>q1A-1.5*diq:
    li = q1A-1.5*diq
else:
    li = base_A["Nota"].min()
```

```python
figura1= px.box(y=base_A["Nota"],title="Box Turma A",points="all")
figura1.show()
```
